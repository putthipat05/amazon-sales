```python

import pandas as pd
import numpy as np

df_amazon = pd.read_csv('amazon.csv')

# =========================
# 1. Drop duplicates
# =========================
df_amazon.drop_duplicates(keep='first', inplace=True)


# =========================
# 2. Product ID
# =========================
df_amazon['product_id'] = (
    df_amazon['product_id']
    .str.strip()
    .str.replace(' ', '_')
    .str.upper()
)


# =========================
# 3. Product name
# =========================
df_amazon['product_name'] = (
    df_amazon['product_name']
    .str.strip()
    .str.title()
)


# =========================
# 4. User name
# =========================
df_amazon['user_name'] = (
    df_amazon['user_name']
    .str.strip()
    .str.capitalize()
)


# =========================
# 5. About product
# =========================
df_amazon['about_product'] = (
    df_amazon['about_product']
    .str.strip()
    .str.replace('\n', ' ', regex=False)
    .str.replace('\r', ' ', regex=False)
    .str.capitalize()
)


# =========================
# 6. Rating
# =========================
df_amazon['rating'] = pd.to_numeric(
    df_amazon['rating'],
    errors='coerce'
)


# =========================
# 7. Rating count
# =========================
df_amazon['rating_count'] = (
    df_amazon['rating_count']
    .str.replace(',', '', regex=False)
)

df_amazon['rating_count'] = pd.to_numeric(
    df_amazon['rating_count'],
    errors='coerce'
).astype('Int64')


# =========================
# 8. Actual price
# =========================
df_amazon['actual_price'] = (
    df_amazon['actual_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df_amazon['actual_price'] = pd.to_numeric(
    df_amazon['actual_price'],
    errors='coerce'
)


# =========================
# 9. Discounted price
# =========================
df_amazon['discounted_price'] = (
    df_amazon['discounted_price']
    .str.replace('₹', '', regex=False)
    .str.replace(',', '', regex=False)
)

df_amazon['discounted_price'] = pd.to_numeric(
    df_amazon['discounted_price'],
    errors='coerce'
)


# =========================
# 10. Discount percentage
# =========================
df_amazon['discount_percentage'] = (
    df_amazon['discount_percentage']
    .str.replace('%', '', regex=False)
)
```

```python
# =========================
# 11. Check data types
# =========================
print(df_amazon.dtypes)


# =========================
# 12. Check missing values
# =========================
print("\nMissing values:")
print(df_amazon.isnull().sum()
```

```python
# =========================
# 13. Upload to MySQL
# =========================
df_amazon.to_sql(
    name="amazon_sales",
    con=engine,
    if_exists="replace",
    index=False
)

print("\nData uploaded successfully!")
```
